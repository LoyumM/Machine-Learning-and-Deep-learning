1. Stochastic Gradient Descent (SGD)

The most basic optimizer. It updates weights by subtracting a fraction of the gradient.

**The Math:**

$$W = W - \eta \cdot \frac{\partial L}{\partial W}$$

Where:

- $W$ = Weights
    
- $\eta$ (eta) = Learning Rate (e.g., 0.01)
    
- $\frac{\partial L}{\partial W}$ = Gradient (dvalues)
    

**The Problem:**

If the learning rate is too low, training takes forever. If it's too high, it overshoots the minimum and diverges. It also gets stuck easily in local minima (shallow valleys in the error surface).

**Python Implementation:**

### 2. SGD with Momentum

Imagine rolling a ball down a hill. SGD is like a ball that stops instantly if the slope disappears. **Momentum** gives the ball mass; it builds up speed (velocity) in the direction of the slope, helping it power through small bumps and flat spots.

**The Math:**

We introduce a `momentum` term (usually 0.9). We calculate a "velocity" ($v$) that accumulates gradients over time.

$$v_{t} = \rho \cdot v_{t-1} + \eta \cdot \text{gradient}$$

$$W = W - v_{t}$$
